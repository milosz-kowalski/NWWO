```python
import random
import time

def symulator_ruletki():
    # Definicja pól ruletki europejskiej (0-36)
    czerwone = {1, 3, 5, 7, 9, 12, 14, 16, 18, 19, 21, 23, 25, 27, 30, 32, 34, 36}
    czarne = {2, 4, 6, 8, 10, 11, 13, 15, 17, 20, 22, 24, 26, 28, 29, 31, 33, 35}
    # 0 to kolor zielony

    bankroll = 1000  # Początkowy stan konta

    print("=" * 45)
    print("      WITAJ W SYMULATORZE RULETKI AI      ")
    print("=" * 45)
    print(f"Początkowy stan konta: {bankroll} zł\n")

    while bankroll > 0:
        print(f"--- AKTUALNY STAN KONTA: {bankroll} zł ---")
        
        # 1. Pobranie wyboru koloru
        wybor = input("Wybierz kolor [c - czerwony, cz - czarny, z - zielony, q - wyjście]: ").strip().lower()
        
        if wybor == 'q':
            print(f"\nDzięki za grę! Odchodzisz ze stołu z kwotą: {bankroll} zł")
            break
            
        if wybor not in ['c', 'cz', 'z']:
            print("Niepoprawny wybór! Wpisz 'c', 'cz', 'z' lub 'q'.\n")
            continue

        # 2. Pobranie stawki
        try:
            stawka = int(input(f"Podaj stawkę (1 - {bankroll}): "))
            if stawka <= 0 or stawka > bankroll:
                print("Niepoprawna kwota stawki!\n")
                continue
        except ValueError:
            print("Wprowadź poprawną liczbę całkowitą!\n")
            continue

        # 3. Animacja kręcenia kołem
        print("\n[  ] Kulka wpada do koła...", end="", flush=True)
        symbol_animacji = ["|", "/", "-", "\\"]
        for i in range(12):
            time.sleep(0.1)
            print(f"\r[{symbol_animacji[i % len(symbol_animacji)]}] Kulka kręci się...", end="", flush=True)
        
        # 4. Losowanie wyniku (0-36)
        wylosowana_liczba = random.randint(0, 36)
        
        if wylosowana_liczba == 0:
            wylosowany_kolor = 'z'
            nazwa_koloru = "ZIELONY (0)"
        elif wylosowana_liczba in czerwone:
            wylosowany_kolor = 'c'
            nazwa_koloru = f"CZERWONY ({wylosowana_liczba})"
        else:
            wylosowany_kolor = 'cz'
            nazwa_koloru = f"CZARNY ({wylosowana_liczba})"

        print(f"\r[!] Wpadło na: {nazwa_koloru}!          \n")

        # 5. Rozliczenie zakładu
        if wybor == wylosowany_kolor:
            # Wygrana
            mnoznik = 14 if wybor == 'z' else 2  # Trafienie zielonego płaci 14x (przy zakładzie na kolor), c/cz płaci 2x
            wygrana = stawka * mnoznik
            zysk = wygrana - stawka
            bankroll += zysk
            print(f" WYGRANA! Trafiłeś kolor! +{zysk} zł (Wygrana: {wygrana} zł)")
        else:
            # Przegrana
            bankroll -= stawka
            print(f" PRZEGRANA! Tracisz {stawka} zł.")

        print()

    if bankroll <= 0:
        print(" BANKRUT! Wygrałeś bilet do domu. Spróbuj szczęścia następnym razem!")

if __name__ == "__main__":
    symulator_ruletki()
```
